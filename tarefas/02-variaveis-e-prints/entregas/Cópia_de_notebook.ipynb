## Parte 1 — Conversor de Temperatura

Você vai converter uma temperatura em Celsius para Fahrenheit e Kelvin.

**Fórmulas:**
- Fahrenheit: `F = C × 9/5 + 32`
- Kelvin: `K = C + 273.15`

```python
# 1. Crie uma variável chamada celsius com um valor de sua escolha
celsius = 25.0

# 2. Calcule o valor em Fahrenheit e armazene em uma variável
fahrenheit = celsius * 9/5 + 32


# 3. Calcule o valor em Kelvin e armazene em uma variável
kelvin = celsius + 273.15


# 4. Exiba os três valores usando print() e f-string
print(f"Temperatura em Celsius: {celsius}°C")
print(f"Temperatura em Fahrenheit: {fahrenheit}°F")
print(f"Temperatura em Kelvin: {kelvin}K")

```

## Parte 2 — Calculadora de Média

Você vai calcular a média de três notas e exibir o resultado formatado.

```python
# 1. Crie três variáveis: nota1, nota2, nota3
nota1 = 8.0
nota2 = 7.5
nota3 = 9.8


# 2. Calcule a média e armazene em uma variável chamada media
media = (nota1 + nota2 + nota3) / 3

# 3. Exiba o resultado usando print() e f-string
print(f"Notas: {nota1} | {nota2} | {nota3}")
print(f"Média Final: {media:.2f}")
```
